```python
results = list(df["result"])
home_wins = 0
away_wins = 0
draws = 0

for result in results:
    if result == "HOME_TEAM":
        home_wins += 1
    elif result == "DRAW":
        draws += 1
    else:
        away_wins += 1

print(f"Home wins: {home_wins}\nAway Wins: {away_wins}\nDraws: {draws}")


```
